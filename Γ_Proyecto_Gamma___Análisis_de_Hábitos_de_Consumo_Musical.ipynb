## Estandarización de categorías musicales

Con el propósito de mejorar la consistencia de la información y garantizar análisis más precisos, se definió un procedimiento para corregir variaciones en la nomenclatura de los géneros musicales. La existencia de múltiples formas de representar una misma categoría puede fragmentar los resultados y afectar la interpretación de las preferencias de los usuarios.

La implementación de un mecanismo de estandarización permite unificar registros equivalentes bajo una categoría común, facilitando comparaciones más confiables, una mejor medición de popularidad por género y una comprensión más precisa de los hábitos de escucha de la audiencia.

```python
# Función para reemplazar los duplicados implícitos
def replace_wrong_values(df, column, wrong_values, correct_value):
    for wrong in wrong_values:
        df[column] = df[column].replace(wrong, correct_value)
    return df
```

## Corrección de inconsistencias en los géneros musicales

Durante la exploración de los datos se identificaron distintas formas de registrar un mismo género musical, lo que podía fragmentar los resultados y afectar la precisión de los análisis. Para evitar que preferencias equivalentes fueran contabilizadas como categorías diferentes, se unificaron las variantes asociadas al género hip hop bajo una única clasificación.

Esta depuración permite representar con mayor fidelidad las preferencias reales de los usuarios, mejorando la calidad de los análisis de popularidad por género y facilitando

```python
# Elimina los duplicados implícitos

df = replace_wrong_values(
    df,
    'genre',
    ['hip', 'hop', 'hip-hop'],
    'hiphop'
)
```

## Creación de una herramienta para el análisis comparativo

Para facilitar el análisis de los hábitos de escucha, se desarrolló una función que permite consultar rápidamente la actividad de los usuarios según una ciudad y un día de la semana específicos. Esta herramienta simplifica la comparación entre grupos y evita repetir procesos de filtrado cada vez que se desea evaluar un nuevo escenario.

Gracias a este enfoque, resulta más sencillo identificar diferencias de comportamiento entre ciudades y detectar patrones de uso en distintos momentos de la semana.

```python
# Declaramos la función number_tracks() con dos parámetros: day= y city=.
def number_tracks(day, city):
    filtered_df = df[df['day'] == day]
    filtered_df = filtered_df[filtered_df['city'] == city]
    return filtered_df['user_id'].count()
```

## Actividad de escucha en Springfield durante los lunes
Al aplicar la función para Springfield los lunes, se obtuvo el mismo resultado observado anteriormente: 15.740 reproducciones. Este valor confirma que la ciudad concentra una parte importante de la actividad registrada en la plataforma al inicio de la semana.

```python
# El número de canciones reproducidas en Springfield el lunes
result = number_tracks('Monday', 'Springfield')
print(result)
```

## Actividad de escucha en Shelbyville durante los lunes

En Shelbyville se registraron 5.614 reproducciones durante los lunes. Aunque la actividad es considerablemente menor que la observada en Springfield, el volumen sigue siendo suficiente para identificar tendencias y patrones de escucha representativos de esta ciudad.

Al comparar ambos resultados, se observa una diferencia clara en el nivel de uso de la plataforma entre las dos ciudades, lo que sugiere comportamientos de consumo distintos que merecen un análisis más detallado en las siguientes etapas del proyecto.

```python
# El número de canciones reproducidas en Shelbyville el lunes
number_tracks('Monday', 'Shelbyville')
```

## Actividad de escucha en Springfield durante los viernes

Durante los viernes, Springfield registró 15.945 reproducciones, una cifra ligeramente superior a la observada los lunes.
